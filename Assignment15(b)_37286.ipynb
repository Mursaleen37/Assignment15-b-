```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\37286\Downloads\student_tea_snacks _1_.csv")
```

# Q1: What is the most preferred tea type?

```python
df.columns
```

```python
print("Q1: Most Preferred Tea Type")
most_preferred_tea = df['tea_type'].value_counts().idxmax()
print(f"The most preferred tea type is: {most_preferred_tea}")
print("\nFull breakdown of tea types:")
print(df['tea_type'].value_counts())
print("-" * 40)
```

# Q2: What is the average number of cups per gender?

```python
print("Q2: Average Number of Cups per Gender")
avg_cups_gender = df.groupby('gender')['cups_per_day'].mean()
print(avg_cups_gender)
print("-" * 40)
```

# Q3: Make a bar chart of snack choices

```python
print("Q3: Generating Bar Chart (Snack Choices)...")
plt.figure(figsize=(8, 5))
df['snack_choice'].value_counts().plot(kind='bar', color='coral', edgecolor='black')
plt.title('Student Snack Choices', fontsize=14)
plt.xlabel('Snack Choice', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
print("-" * 40)
```

# Q4: Compare tea preference by age group

```python
print("Q4: Compare Tea Preference by Age Group")
tea_by_age = pd.crosstab(df['age_group'], df['tea_type'])
print(tea_by_age)
print("-" * 40)
```
